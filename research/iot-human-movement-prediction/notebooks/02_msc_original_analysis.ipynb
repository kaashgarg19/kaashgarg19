# 02 — MSc foundation analysis

This notebook reconstructs the research logic of the original MSc project: environmental IoT telemetry → preprocessing → visualisation → human-movement classification. It is a **reproducible reconstruction**, not a claim that every line is the exact 2021 source code.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
sys.path.append(str(Path.cwd().parent / 'src'))
from data import load_dataset, FEATURES, TARGET, prepare_time

df = prepare_time(load_dataset(Path('../data/raw/iotdata.csv')))
```

```python
# Basic descriptive analysis
display(df[FEATURES + [TARGET]].describe().T)
print('Class prevalence:', df[TARGET].mean())
```

```python
# Original MSc-style telemetry visualisation
fig, axes = plt.subplots(3, 2, figsize=(12, 10))
for ax, feature in zip(axes.ravel(), FEATURES):
    ax.plot(df['datetime_utc'], df[feature], linewidth=0.5)
    ax.set_title(feature)
    ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
```

## Boundary between MSc and extension

The MSc foundation establishes the prediction problem and exploratory analysis. The later research extension adds controlled random evaluation, chronological evaluation, unseen-device testing, feature ablation, threshold sensitivity and uncertainty analysis.
